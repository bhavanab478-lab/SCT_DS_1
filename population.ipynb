```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("population.csv")

# Display first rows
print(df.head())

# -------- Bar Chart --------
plt.figure(figsize=(10,5))
plt.bar(df['Country Name'], df['2022 Population'], color='blue')
plt.title("Population Distribution by Country")
plt.xlabel("Country")
plt.ylabel("Population")
plt.xticks(rotation=90)
plt.show()


# -------- Histogram --------
plt.figure(figsize=(8,5))
plt.hist(df['2022 Population'], bins=10, color='green', edgecolor='black')
plt.title("Histogram of Population Distribution")
plt.xlabel("Population")
plt.ylabel("Frequency")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

# Upload Titanic dataset
uploaded = files.upload()

# Read dataset
df = pd.read_csv("train.csv")   # change file name if needed

# First 5 rows
print(df.head())

# Dataset info
print(df.info())

# Check missing values
print(df.isnull().sum())

# Data Cleaning
# Fill missing Age values with mean
df['Age'].fillna(df['Age'].mean(), inplace=True)

# Fill missing Embarked values with mode
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

# Drop Cabin column (many missing values)
df.drop('Cabin', axis=1, inplace=True)

print("Missing values after cleaning:")
print(df.isnull().sum())

# -------- EDA --------

# Survival count
sns.countplot(x='Survived', data=df)
plt.title("Survival Count")
plt.show()

# Survival by gender
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Survival by Gender")
plt.show()

# Age distribution
plt.hist(df['Age'], bins=20, edgecolor='black')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

# Passenger class distribution
sns.countplot(x='Pclass', data=df)
plt.title("Passenger Class Distribution")
plt.show()

# Correlation heatmap
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
